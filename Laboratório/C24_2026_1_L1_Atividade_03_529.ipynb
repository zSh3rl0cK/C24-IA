## <font color='green'>**2.5 Identificação de Problemas nos Dados**</font>

### <font color='cyan'> Q16: Inspecione o conjuntos de dados para verificar se há:</font>
- #### <font color='cyan'> Dados Ausentes;</font>
- #### <font color='cyan'> Valores Atípicos (Outliers);</font>
- #### <font color='cyan'> Valores Duplicados;</font>
### <font color='cyan'> Reporte sua(s) descoberta(s) a seguir.</font>

```python
# P/ valores duplicados
# Quantas linhas duplicadas existem?
print("Valores duplicados: ", df.duplicated().sum())

# P/ valores ausentes
# Quantas linhas de dados ausentes tem?
print("Valores ausentes: ", df.isna().sum().sum())

# P/ Outliers
df_numeric_columns = df.select_dtypes(include=['number'])

outliers_por_coluna = []
df_sem_outliers_iqr = df_numeric_columns.copy()
for col in df_numeric_columns:
    Q1 = df_sem_outliers_iqr[col].quantile(0.25)
    Q3 = df_sem_outliers_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lim_inf = Q1 - 1.5 * IQR
    lim_sup = Q3 + 1.5 * IQR

    n_outliers = len(df_numeric_columns[
        (df_numeric_columns[col] < lim_inf) |
        (df_numeric_columns[col] > lim_sup)
    ])

    outliers_por_coluna.append({'Colunas': col, 'Outliers': n_outliers})

df_outliers = pd.DataFrame(outliers_por_coluna)
print("Outliers por coluna")
display(df_outliers)
```

### <font color='cyan'> Q17: Caso problemas tenham sido encontrados, proponha soluções para eles.</font>
### <font color='cyan'> Exiba o conjunto de dados antes e depois de aplicar as soluções, enfatizando as mudanças.</font>

```python
# Solução para as duplicatas
print("Removendo as duplicatas\n")
df_tratado = df.drop_duplicates()

# Solução para valores ausentes não necessária pois nenhum foi encontrado
print("Solução para valores ausentes não necessária pois nenhum foi encontrado\n")

# Solução para outliers
explicacao = (
    "Em Cybersecurity, outliers podem ser sinais de tipos de  ataques novos.\n"
    " havendo então um risco, pois a remoção pode causar a perda de padrões críticos de intrusão.\n"
    "Para exemplificar um possível tratamento (no caso, a remoção dos outliers)\n"
    "criarei uma cópia isolada (df_sem_outliers_iqr) para fins de comparação\n"
    "e visualização estatística, sem afetar o conjunto de dados original."
)

print(explicacao + "\n")

df_sem_outliers_iqr = df_tratado.copy()

cols_numericas = df_sem_outliers_iqr.select_dtypes(include=['number']).columns

for col in cols_numericas:
    Q1 = df_sem_outliers_iqr[col].quantile(0.25)
    Q3 = df_sem_outliers_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lim_inf = Q1 - 1.5 * IQR
    lim_sup = Q3 + 1.5 * IQR

    # Filtrando o DataFrame: Mantemos apenas o que está dentro dos limites
    # Aplicamos o filtro sucessivamente para cada coluna
    df_sem_outliers_iqr = df_sem_outliers_iqr[
        (df_sem_outliers_iqr[col] >= lim_inf) &
        (df_sem_outliers_iqr[col] <= lim_sup)
    ]

print(f"Linhas antes do tratamento: {len(df_tratado)}")
print(f"Linhas após remover outliers: {len(df_sem_outliers_iqr)}")
print(f"Total de linhas removidas: {len(df_tratado) - len(df_sem_outliers_iqr)}")
```

### <font color='cyan'> Q19: Calcule e exiba o Coeficiente de Correlação de Pearson ($r$) entre elas.</font>

```python
correlacao = df_tratado['fwd_pkts_tot'].corr(df_tratado['bwd_pkts_tot'])

print(f"Coeficiente de correlação de Pearson (r):  {correlacao:.4f}")

plt.scatter(df_tratado['fwd_pkts_tot'], df_tratado['bwd_pkts_tot'])
plt.xlabel('fwd_pkts_tot')
plt.ylabel('bwd_pkts_tot')
plt.grid(True)
plt.show()
```
